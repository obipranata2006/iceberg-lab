```python
# BEFORE compaction: ukur kondisi file
spark.sql("""
SELECT count(*) AS data_file_count,
       sum(record_count) AS total_records,
       round(sum(file_size_in_bytes)/1024/1024, 2) AS total_mb
FROM demo.undiknas.student_activity.files
""").show()

# Query uji logis
spark.sql("""
SELECT program, count(*) AS n, round(avg(score),2) AS avg_score
FROM demo.undiknas.student_activity
GROUP BY program ORDER BY program
""").show()

# Compaction
spark.sql("""
CALL demo.system.rewrite_data_files(
  table => 'undiknas.student_activity',
  options => map('target-file-size-bytes','134217728')
)
""").show()

# AFTER compaction: ukur lagi dengan query yang sama
spark.sql("""
SELECT count(*) AS data_file_count,
       sum(record_count) AS total_records,
       round(sum(file_size_in_bytes)/1024/1024, 2) AS total_mb
FROM demo.undiknas.student_activity.files
""").show()

spark.sql("""
SELECT program, count(*) AS n, round(avg(score),2) AS avg_score
FROM demo.undiknas.student_activity
GROUP BY program ORDER BY program
""").show()
```
